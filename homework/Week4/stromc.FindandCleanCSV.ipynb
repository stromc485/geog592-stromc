```python
# Importing csv
import csv

# Read and inspect the raw dataset using csv.reader
file = "Hospitals_NC.csv"

with open(file, "r", encoding="utf-8-sig") as file: # Open the CSV file in read mode with UTF-8 encoding
    reader = csv.reader(file) # This creates a CSV reader object to read the file
    rows = list(reader)

# Verify data structure by printing the first 10 rows
for row in rows[:10]: 
    print(row) 
```

```python
# Extract the header and data rows
header = rows[0]
data_rows = rows[1:]
```

```python
# Locate key column indices dynamically for spatial join processing
zip_idx = header.index("fzip")
county_idx = header.index("fcounty")

cleaned_rows = [header]
```

```python
# Process each row
for row in data_rows:
    # Strip whitespace from every cell
    cleaned_row = [cell.strip() for cell in row]

    # Format 5-digit ZIP codes (fzip) for ZCTA vector joins
    if cleaned_row[zip_idx]:
        cleaned_row[zip_idx] = cleaned_row[zip_idx][:5].zfill(5)

    # Standardize County names (fcounty) to Title Case for Shapefile/GeoJSON joins
    if cleaned_row[county_idx]:
        cleaned_row[county_idx] = cleaned_row[county_idx].title()

    cleaned_rows.append(cleaned_row)
```

```python

output_file = "cleaned_Hospitals_NC.csv" # Output file name for the cleaned data
with open(output_file, mode="w", newline="", encoding="utf-8") as outfile:
    writer = csv.writer(outfile)
    writer.writerows(cleaned_rows)

print("Cleaned CSV saved as:", output_file) # Print statement 
```
