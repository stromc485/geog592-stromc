```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
```

```python
CAFOs = gpd.read_file("/Users/Charlotte/Desktop/SeniorYear/GEOG_592/geog592-stromc/geog592-stromc/homework/Week07/data/List_Permitted_Animal_Facilities20260423.csv")
NC_county_boundaries = gpd.read_file("/Users/Charlotte/Desktop/SeniorYear/GEOG_592/geog592-stromc/geog592-stromc/homework/Week07/data/North_Carolina_State_and_County_Boundary_Polygons.geojson")
```

```python
# Cleaning For Only Hog CAFOS 
Hog_CAFOs = CAFOs[CAFOs["Regulated Operation"] == "Swine"]
Hog_CAFOs
```

```python
# Converting NC County Boundaries to EPSG:6543
NC_county_boundaries_6543 = NC_county_boundaries.to_crs(epsg=6543)
NC_county_boundaries_6543.crs
```

```python
# Joining point and polygon data
df1 = pd.DataFrame(Hog_CAFOs[['Permit Number', 'Facility Name', 'Combined Owner',
       'Regulated Operation', 'Permit Type', 'Regulated Activity',
       'Allowable Count', 'Number Of Lagoons', 'Issued Date', 'Effective Date',
       'Expiration Date', 'Admin Region', 'County Name', 'Location Lat Num',
       'Location Long Num', 'Address 1', 'Address 2', 'City', 'State', 'Zip',
       'field_21', 'field_22']])
df2 = pd.DataFrame(NC_county_boundaries[['OBJECTID', 'County', 'FIPS', 'Rec_Survey', 'NCGS_url', 'ck_date',
       'Shape__Area', 'Shape__Length', 'GlobalID', 'geometry']])


# Perform a regular inner join on the common column
joined = gpd.GeoDataFrame(pd.merge(df1, df2, 
                     left_on="County Name",
                     right_on = "County",
                       how="inner"),
                       geometry="geometry",
                       crs=NC_county_boundaries_6543.crs)

print(joined)
```

```python
# Number of hog CAFOs per NC county
cafo_counts = joined.groupby("County").size()

cafo_counts = cafo_counts.reindex(
    NC_county_boundaries["County"].unique(),
    fill_value=0
)
cafo_counts
```

Creating an atlas using a loop

```python
# Importing necessary libraries
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
```

```python
# Creating PDF 
# Using AI:
with PdfPages("NC_CAFO_Atlas.pdf") as pdf:

    for index, county in NC_county_boundaries.iterrows():

        county_name = county["County"]

        # Get CAFO count for the county
        cafo_count = cafo_counts[county_name]

        # Select CAFOs in the county
        county_cafos = joined[joined["County"] == county_name]

        # Create figure
        fig, ax = plt.subplots(figsize=(8, 8))

        # Plot county boundary
        gpd.GeoDataFrame(
            [county],
            geometry="geometry",
            crs=NC_county_boundaries.crs
        ).plot(ax=ax, color="lightgrey", edgecolor="black")

        # Plot CAFO locations
        county_cafos.plot(ax=ax, color="red", markersize=10)

        # Create sentence using string concatenation
        sentence = (
            "There are " + str(cafo_count) +
            " CAFOs in " + county_name + " County."
        )

        # Add title and sentence
        ax.set_title(county_name + " County, North Carolina")
        ax.text(0.5, -0.05, sentence,
                transform=ax.transAxes,
                ha="center",
                fontsize=11)

        ax.set_axis_off()

        # Save page to PDF
        pdf.savefig(fig, bbox_inches="tight")
        plt.close(fig)
```
