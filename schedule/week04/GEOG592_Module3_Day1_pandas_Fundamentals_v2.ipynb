## A Short Introduction to Classes and Objects

You do **not** need to learn object-oriented programming today. You just need the basic concepts to understand code such as:

```python
df.head()
df.shape
df["Data_Value"].mean()
```

Here are the definitions.
- **Class** : a template that defines the kind of data and behavior needed to create an *object*.
- **Object** :  one actual instance created from a class.  Objects in a class have *attributes* and *methods.*
- **Attribute** : refers to specific data that belongs to an object. 
- **Method** : refers to a function that belongs to the object; usually all objects of the same class have the same methods (but the methods' behaviour can depent on the specific attributes)

### Example of a class

The `Point` class defined below represents a point. It stores two attributes (`x` and `y`) and has two methods.

You are **not expected to be able to write a class definition like this on the assessment**. We are using it only to make dot notation visible.

````python
class Point:
    def __init__(self, the_x_value, the_y_value):
        self.x = the_x_value ## here we are creating the x attribute
        self.y = the_y_value ## here we are creating the y attribute

    def show_coordinates(self): ## here we are creating the show_coordinates method
        print(self.x, self.y)

    def move(self, dx, dy): ## here we are creating the move method
        self.x = self.x + dx
        self.y = self.y + dy
````

The `Point` class has two attributes.

In the next code block, we create two objects from the same `Point` class.

````python
point_a = Point(100, 200)
point_b = Point(300, 450)

print(point_a.x)
print(point_b.x)
````

In this example:

| Python element | What is it? |
|---|---|
| `Point` | a class |
| `point_a` | an object |
| `point_b` | another object |
| `x`, `y` | attributes |
| `show_coordinates()` | a method |
| `move()` | another method |


#### Dot notation
The **dot** can be read as "something that belongs to this object."

````python
point_a.show_coordinates()

point_a.move(10, 20)
point_a.show_coordinates()
````

### Attribute vs. method

Look at the difference between these two statements.

```python
point_a.x
point_a.show_coordinates()
```
`x` is an **attribute**. It stores information.
`show_coordinates()` is a **method** that performs an action.

A useful visual clue is that methods are **called with parentheses**.

## Selecting Rows by Position with `iloc`

`iloc` selects rows and columns by integer position.

```python

```

returns the first row.

````python
df.iloc[0]
````

````python
df.iloc[1:4][["LocationName", "Data_Value"]]
````

Note from the above can also use `iloc[]` with slice notation to select a range of rows. In this case `1:4` means *include index 1 and everything is between 1 and 4, excluding 4.* 

## Boolean Comparisons

A comparison against a pandas column produces a Boolean Series.

````python
df["Year"] == "2023"
````

The result contains `True` or `False` for every row.

We can store this Boolean Series in a variable called a **mask**.

````python
valid_year = df["Year"] == "2023"
valid_year.head()
````

## Filtering Rows

Use the Boolean mask inside the DataFrame brackets.

````python
print('length before mask', len(df))
print('length of mask is' ,len(valid_year) ,' and needs to be the same as the length of the df that you want to mask')

clean_year = df[valid_year]

print('length of our new df after using the mask:', len(clean_year))
print('since we had', len(valid_year[valid_year == True]), 'True values')
clean_year.head()
````

Here I am going to mask using information from another column. I am going to mask all the rows that have a county name that starts with an A. 
Think of masking like a fast way of doing a looped `if` statement. 

````python
mask = df["LocationName"].str.startswith("A", na=False)
df2 = df[mask]
df2
````

One of the most common masking errors is trying to mask a DataFrame using a mask of the wrong dimmension or with different indexing. 

## Multiple Conditions

Use `&` for AND when combining pandas Boolean conditions. Put each condition in parentheses.

````python
selected = df[
    (df["Year"] == "2023") &
    (df["StateAbbr"] == "NC")
]

selected[["LocationName", "Data_Value"]].head()
````

Later we will see the same idea again in NumPy and GeoPandas. **Boolean filtering is one of the most transferable ideas in the course.**
