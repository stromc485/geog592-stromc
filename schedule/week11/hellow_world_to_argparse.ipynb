```python
import sys

def say_hello(name):
    print(f"Hello, {name}!")

def main():
    name = sys.argv[1] # will return an error because the list is empty if no argument is passed
    say_hello(name)

if __name__ == "__main__":
    main()
```

```python
import argparse

def say_hello(name, times):
    for i in range(times):
        print(f"Hello, {name}!")

def main():
    parser = argparse.ArgumentParser(
        description="A program that greets someone."
    )
    parser.add_argument(
        "name",
        help="Name of the person to greet.")

    parser.add_argument(
        "--times",
        type=int,
        default=1,
        help="Number of times to greet (default: 1)")

    args = parser.parse_args()
    say_hello(args.name, args.times)

if __name__ == "__main__":
    main()
```
