### Geoprocessing Technique #2: Buffering

I want to find how many swine CAFOs are within one mile of each other in North Carolina. 
    
    To do this, I am doing a proximity analysis by buffering around each hog CAFO

```python
# Create one mile buffers
buffers = gdf_cafos_6543.copy()
buffers['geometry'] = buffers.geometry.buffer(5280)
```

```python

# Now, perform a spatial join to find CAFOs inside each buffer
joined = gpd.sjoin(
    buffers,
    gdf_cafos_6543,
    how='left',
    predicate='contains'
)
```

```python
# Count CAFOs within one mile of each CAFO
counts = (
    joined.groupby('Permit Number_left')
    .size()
    .reset_index(name='CAFOs_within_1_mile')
)
```

```python
# Add counts back to original CAFO layer
gdf_cafos_6543 = gdf_cafos_6543.merge(
    counts,
    left_on='Permit Number',
    right_on='Permit Number_left',
    how='left'
)
```

```python
gdf_cafos_6543['CAFOs_within_1_mile_mile'] = (
    gdf_cafos_6543['CAFOs_within_1_mile'].fillna(0)
)

# Remove the CAFO itself from its own count
gdf_cafos_6543['CAFOs_within_1_mile'] = (
    gdf_cafos_6543['CAFOs_within_1_mile'] - 1
)
```

```python
# NOW finding only the CAFOs that are within 1 mile of another CAFO
overlapping_cafos = gdf_cafos_6543[
    gdf_cafos_6543['CAFOs_within_1_mile'] > 0
]
len(overlapping_cafos)
```

    There are 1,414 hog CAFOs within one mile of another CAFO in North Carolina.

```python
# Plot the CAFOs with the count of other CAFOs within 1 mile
fig, ax = plt.subplots(figsize=(10,10))
overlapping_cafos.plot(ax=ax, markersize=5, c="red")
gdf_cafos_6543.plot(ax=ax, markersize=0.5, c="black")
ax.set_title("CAFOs with Overlapping Buffers of 1-mile and Original Points")
plt.show()
```

### Geoprocessing Technique #4: Tabular Join

Using a tabular join to display how CAFOs are distributed across North Carolina counties, specifically looking at how  CAFO density relates to county characteristics.

```python
print(NC_county_boundaries_6543.columns)
print(Hog_CAFOs.columns)
```

```python
df1 = pd.DataFrame(Hog_CAFOs[['Permit Number', 'Facility Name', 'Combined Owner',
       'Regulated Operation', 'Permit Type', 'Regulated Activity',
       'Allowable Count', 'Number Of Lagoons', 'Issued Date', 'Effective Date',
       'Expiration Date', 'Admin Region', 'County Name', 'Location Lat Num',
       'Location Long Num', 'Address 1', 'Address 2', 'City', 'State', 'Zip',
       'field_21', 'field_22']])
df2 = pd.DataFrame(NC_county_boundaries[['OBJECTID', 'County', 'FIPS', 'Rec_Survey', 'NCGS_url', 'ck_date',
       'Shape__Area', 'Shape__Length', 'GlobalID', 'geometry']])


# Perform a regular inner join on the common column
joined = gpd.GeoDataFrame(pd.merge(df1, df2, 
                     left_on="County Name",
                     right_on = "County",
                       how="inner"),
                       geometry="geometry",
                       crs=NC_county_boundaries_6543.crs)

print(joined)
```

```python
# Count CAFOs per county
cafo_count_by_county = Hog_CAFOs['County Name'].value_counts().reset_index()
cafo_count_by_county.columns = ['County Name', 'CAFO_count']

# Merge the counts into joined
joined = joined.merge(cafo_count_by_county, on='County Name', how='left')

# Calculate CAFOs per 100 sq miles
NC_county_boundaries_6543['County_area_sqmi'] = NC_county_boundaries_6543.geometry.area / (5280**2)
joined['County_area_sqmi'] = joined.merge(
    NC_county_boundaries_6543[['County', 'County_area_sqmi']], 
    left_on='County', right_on='County', how='left'
)['County_area_sqmi']

joined['CAFOs_per_100_sqmi'] = (
    joined['CAFO_count'] /
    joined['County_area_sqmi']
) * 100

print(joined[['County Name', 'CAFO_count', 'County_area_sqmi', 'CAFOs_per_100_sqmi']].head())
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

# Merge the density data into the county boundaries
county_density = joined[['County', 'CAFOs_per_100_sqmi']].drop_duplicates()
NC_county_boundaries_6543_plot = NC_county_boundaries_6543.merge(
    county_density, on='County', how='left'
)

NC_county_boundaries_6543_plot.plot(
    ax=ax,
    column='CAFOs_per_100_sqmi',
    cmap='YlOrRd',
    edgecolor='black',
    legend=True,
    legend_kwds={'label': 'CAFOs per 100 sq miles'}
)

ax.set_title('Swine CAFO Density by NC County')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
plt.tight_layout()
plt.show()
```

### Geoprocessing Technique #5: Spatial Join

I want to find out how the Cape Fear River Basin compares with the other river basins in North Carolina in numbers of swine CAFOs

    Let's start with calculating swine CAFOs in the 16 other river basins... using a for loop! 

```python
for Basins in River_basins_6543['Basin'].unique():
    # Select the current basin
    current_basin = River_basins_6543[River_basins_6543['Basin'] == Basins]
    
    # Perform a spatial join to find all CAFOs that are within the current basin
    cafos_in_basin = gpd.sjoin(gdf_cafos_6543, current_basin, 
        how='inner',
        predicate='within'
    )
    
    # Print the number of CAFOs in the current basin
    print(f"Number of CAFOs in {Basins} Basin: {len(cafos_in_basin)}")
```
