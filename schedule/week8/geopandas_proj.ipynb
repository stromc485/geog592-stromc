Cleaning CAFOs dataset to only be hog CAFOs: 

```python
Hog_CAFOs = CAFOs[CAFOs["Regulated Operation"] == "Swine"]
Hog_CAFOs
```

Converting Both to EPSG:6543 (NAD83 / North Carolina (ft/us)):

    This is a better CRS because it is measured in feet rather than in decimal degrees (which EPSG:4326 is in)

```python
River_basins_6543 = River_basins.to_crs(epsg=6543)
NC_county_boundaries_6543 = NC_county_boundaries.to_crs(epsg=6543)
```

```python
print(River_basins_6543.crs)
print(NC_county_boundaries_6543.crs)
```

Performing a Spatial Join of NC_county_boundaries and River_basins

```python
Joined_layer = gpd.sjoin(River_basins_6543, NC_county_boundaries_6543)


# Create a matplotlib figure and axis
fig, ax = plt.subplots(figsize=(12, 12))

# Plot the underlying counties in light gray
NC_county_boundaries_6543.plot(ax=ax, color='none', edgecolor='black', linewidth=1, label='Counties')

# Plot the joined layer on top, colored by a specific attribute (e.g., Basin Name)
Joined_layer.plot(ax=ax, column='Basin', cmap='tab20', alpha=0.5, legend=True)

plt.title("River Basins Joined with NC Counties")
plt.show()

```

    I like this but it's hard to see the county boundaries, let's join with NC_county_boundaries as the left literal (so it overlays River_basins)

```python
Joined_layer2 = gpd.sjoin(NC_county_boundaries_6543, River_basins_6543)
# create a matplotlib figure and axis
fig, ax = plt.subplots(figsize=(12, 12))

# plot the underlying riverbasins by basin attribute
River_basins_6543.plot(ax=ax, column='Basin', cmap='tab20', alpha=1, legend=True)
Joined_layer2.plot(ax=ax, color='none', edgecolor='white', linewidth=1.2, label='Counties')
plt.title("NC Counties Joined with River Basins")
plt.show()
```

Now, time to perform a join between the joined_layer2 and swine CAFOs 

```python
# Let's look at joined_layer2 data to see what attributes we have available for joining
Joined_layer2
```

```python
# Let's look at Hog_CAFOs data to see what attributes we have available for joining
Hog_CAFOs
```

Okay, I see that Hog_CAFOs has a "County Name" column, while joined_layer2 has a "County" column... let's try that for the join. 


```python
df1 = pd.DataFrame(Hog_CAFOs[['Permit Number', 'Facility Name', 'Combined Owner',
       'Regulated Operation', 'Permit Type', 'Regulated Activity',
       'Allowable Count', 'Number Of Lagoons', 'Issued Date', 'Effective Date',
       'Expiration Date', 'Admin Region', 'County Name', 'Location Lat Num',
       'Location Long Num', 'Address 1', 'Address 2', 'City', 'State', 'Zip',
       'field_21', 'field_22']])
df2 = pd.DataFrame(Joined_layer2[['OBJECTID', 'County', 'FIPS', 'Rec_Survey', 'NCGS_url', 'ck_date',
       'Shape__Area', 'Shape__Length', 'GlobalID_left', 'geometry',
       'index_right', 'FID', 'Basin', 'Sq_Miles', 'Acres', 'Name', 'PlanLink',
       'GlobalID_right']])


# Perform a regular inner join on the common column
merged_df = gpd.GeoDataFrame(pd.merge(df1, df2, 
                     left_on="County Name",
                     right_on = "County",
                       how="inner"),
                       geometry="geometry",
                       crs=Joined_layer2.crs)

print(merged_df)
```

Creating a point GeoDataFrame from the CAFO coordinates

```python
gdf_cafos = gpd.GeoDataFrame(
    Hog_CAFOs,
    geometry=gpd.points_from_xy(Hog_CAFOs["Location Long Num"], Hog_CAFOs["Location Lat Num"]),
    crs="EPSG:4326"
)
```

Reproject gdf_cafos to match the map CRS

```python
gdf_cafos_6543 = gdf_cafos.to_crs(epsg=6543)
gdf_cafos_6543.crs # ensure it worked 
```

Now, plot the map


```python
from matplotlib import colormaps

fig, ax = plt.subplots(figsize=(12, 12))

# Basins 
River_basins_6543.plot(ax=ax, column='Basin', cmap='tab20', alpha=1)
# Counties 
Joined_layer2.plot(ax=ax, color='none', edgecolor='gray', linewidth=1.2)
# CAFOs
gdf_cafos_6543.plot(ax=ax, color='darkred', markersize=10, label='Swine CAFOs')
plt.title("Swine CAFOs joined to River Basins and Counties") 
plt.show()
```

Now, map swine CAFOs within a specific river basin. Let's choose CPF (Cape Fear.)

```python
# Select the Cape Fear River Basin from the River_basins_6543 GeoDataFrame
CPF_RB = River_basins_6543[(River_basins_6543['Basin'] == 'CPF') & (River_basins_6543['Name'] == 'Cape Fear')]
```

```python
# Now, perform a spatial join to find all CAFOs that are within the Cape Fear River Basin 
cafos_in_cpf = gpd.sjoin(gdf_cafos_6543, CPF_RB, 
    how='inner',
    predicate='within'
)
len(cafos_in_cpf)
```

    There are 1,168 CAFO points within the Cape Fear basin geometry. 

Now, visualizing the number of swine CAFOs in the Cape Fear River Basin:

```python
fig, ax = plt.subplots(figsize=(12, 12))
# Basins 
CPF_RB.plot(ax=ax, column='Basin', color='lightgreen', alpha=1, edgecolor='black', linewidth=1, label='Cape Fear River Basin')
# CAFOs
cafos_in_cpf.plot(ax=ax, color='darkred', markersize=10, label='Swine CAFOs')
plt.title("Swine CAFOs in the Cape Fear Basin joined to River Basins and Counties")
plt.legend()
```

Just for fun, let's map the swine CAFOs in the Cape Fear River basin with NC Counties

```python
fig, ax = plt.subplots(figsize=(12, 12))

# Basins 
CPF_RB.plot(ax=ax, column='Basin', color='Lightgreen', alpha=1, edgecolor='black', linewidth=1, label='Cape Fear River Basin') 
# CAFOs in the Cape Fear Basin
cafos_in_cpf.plot(ax=ax, color='darkred', markersize=10, label='Swine CAFOs', alpha=1)
# Counties
NC_county_boundaries_6543.plot(ax=ax, color='none', edgecolor='gray', alpha = 0.3, linewidth=1.2)
plt.title("Swine CAFOs in the Cape Fear Basin joined to River Basins and Counties") 
plt.legend()
```

How does the Cape Fear River Basin compare to the others in North Carolina in numbers of swine CAFOs? 

    Let's start with calculating swine CAFOs in the 16 other river basins... can I create a loop for this? 

```python
for Basins in River_basins_6543['Basin'].unique():
    # Select the current basin
    current_basin = River_basins_6543[River_basins_6543['Basin'] == Basins]
    
    # Perform a spatial join to find all CAFOs that are within the current basin
    cafos_in_basin = gpd.sjoin(gdf_cafos_6543, current_basin, 
        how='inner',
        predicate='within'
    )
    
    # Print the number of CAFOs in the current basin
    print(f"Number of CAFOs in {Basins} Basin: {len(cafos_in_basin)}")
```

Now, let's create a bar chart to visualize our findings:

```python
River_Basin_Names = ['BRD', 'CAT', 'CHO', 'CPF', 'FBR', 'HIW', 'LBR', 'LTN', 'NEU', 'NEW', 'PAS', 'ROA', 'SAV', 'TAR', 'WAT', 'WOK', 'YAD']
Num_Hog_CAFOs = [0, 1, 39, 1168, 0, 0, 204, 0, 487, 0, 19, 24, 0, 96, 0, 39, 40]

# 2. Create the bar chart
plt.bar(River_Basin_Names, Num_Hog_CAFOs, color='skyblue', width=1)

# 3. Add titles and labels
plt.title('Number of Hog CAFOs by River Basin')
plt.xlabel('River Basins')
plt.xticks(fontsize=6)  # Rotate x-axis labels for better readability
plt.ylabel('Number of CAFOs')
plt.grid(axis='y', linestyle='--', alpha=0.7)
# 4. Display the graph
plt.show()
```

We can see that the Cape Fear River Basin has significantly more swine CAFOs than any other river basin, followed by the Neuse River Basin and the Lumber River Basin.
